```python
import pandas as pd
```

```python
qg = pd.read_csv("query_generation.csv")
```

```python
import ast

total_query_count = 0
query_list = []
query_list_time = []

for index, row in qg.iterrows():
    queries = ast.literal_eval(row["Queries"])  # convert string → list
    OriginalTimestamps = ast.literal_eval(row["OriginalTimestamps"])  # convert string → list
    # print(queries)
    # print(type(queries))
    # print(len(queries))
    # print(queries[0])
    # break
    total_query_count += len(queries)
    query_list_time.extend(OriginalTimestamps)
    query_list.extend(queries)
```
